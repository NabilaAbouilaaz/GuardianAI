```python
import thrember, numpy as np, joblib, json, time, os
import lightgbm as lgb
from sklearn.metrics import roc_curve, roc_auc_score, confusion_matrix, classification_report

DATA = "/content/ember2024"
OUT  = "/content/drive/MyDrive/GuardianAI/models"
os.makedirs(DATA, exist_ok=True)
t0 = time.time()

# 1. Téléchargement
thrember.download_dataset(DATA, file_type="Dot_Net")
thrember.download_dataset(DATA, split="challenge")
print(f"[{(time.time()-t0)/60:.1f} min] téléchargement terminé")

# 2. Vectorisation
thrember.create_vectorized_features(DATA)
print(f"[{(time.time()-t0)/60:.1f} min] vectorisation terminée")

# 3. Chargement + filtrage des non-étiquetés (-1)
X_train, y_train = thrember.read_vectorized_features(DATA, subset="train")
X_test,  y_test  = thrember.read_vectorized_features(DATA, subset="test")
X_chal,  y_chal  = thrember.read_vectorized_features(DATA, subset="challenge")

idx = np.where((y_train == 0) | (y_train == 1))[0]
if len(idx) > 300000:                      # limite mémoire Colab
    idx = np.sort(np.random.default_rng(42).choice(idx, 300000, replace=False))
Xtr, ytr = X_train[idx], y_train[idx]

it = np.where((y_test == 0) | (y_test == 1))[0]
Xte, yte = X_test[it], y_test[it]

print("train:", Xtr.shape, np.bincount(ytr.astype(int)))
print("test :", Xte.shape, np.bincount(yte.astype(int)))
print("challenge:", X_chal.shape)

# 4. Entraînement
model = lgb.LGBMClassifier(n_estimators=300, num_leaves=256,
                           learning_rate=0.05, n_jobs=-1, random_state=42)
model.fit(Xtr, ytr)
print(f"[{(time.time()-t0)/60:.1f} min] entraînement terminé")

# 5. Évaluation + seuil sous contrainte 2% de faux positifs
p_test = model.predict_proba(Xte)[:, 1]
fpr, tpr, thr = roc_curve(yte, p_test)
ok = np.where(fpr <= 0.02)[0]
seuil = float(thr[ok[-1]])

print("\n--- Seuil 0.5 (par défaut) ---")
print(classification_report(yte, (p_test >= 0.5).astype(int),
                            target_names=['benin', 'malveillant'], digits=4))
print("ROC-AUC:", roc_auc_score(yte, p_test))

print(f"\n--- Seuil optimisé : {seuil:.4f} ---")
yp = (p_test >= seuil).astype(int)
cm = confusion_matrix(yte, yp)
print(cm)
print(f"Faux positifs : {cm[0,1]/cm[0].sum()*100:.2f}%  |  Détection : {cm[1,1]/cm[1].sum()*100:.2f}%")

p_chal = model.predict_proba(X_chal)[:, 1]
det_chal = float((p_chal >= seuil).mean())
print(f"Détection sur le challenge set (malwares évasifs) : {det_chal*100:.2f}%")

# 6. Sauvegarde sur Drive
joblib.dump(model, f"{OUT}/lightgbm_ember2024_v1.joblib")
json.dump({"seuil": seuil, "roc_auc": float(roc_auc_score(yte, p_test)),
           "fpr": float(cm[0,1]/cm[0].sum()), "detection": float(cm[1,1]/cm[1].sum()),
           "detection_challenge": det_chal, "n_features": int(Xtr.shape[1])},
          open(f"{OUT}/metrics_ember2024_v1.json", "w"), indent=2)
!ls -lh /content/drive/MyDrive/GuardianAI/models/
print(f"\nTOTAL : {(time.time()-t0)/60:.1f} minutes")
```

```python
import os, gc, time, json, joblib
import numpy as np, lightgbm as lgb
from sklearn.metrics import roc_curve, roc_auc_score, confusion_matrix, classification_report

DATA = "/content/ember2024"
OUT  = "/content/drive/MyDrive/GuardianAI/models"
if not os.path.exists('/content/drive/MyDrive'):
    from google.colab import drive; drive.mount('/content/drive')
os.makedirs(OUT, exist_ok=True)

import thrember
t0 = time.time()

# --- Entraînement (jeu de test PAS encore chargé, pour économiser la RAM)
X_train, y_train = thrember.read_vectorized_features(DATA, subset="train")
idx = np.where((y_train == 0) | (y_train == 1))[0]
idx = np.sort(np.random.default_rng(42).choice(idx, 200000, replace=False))
Xtr = np.ascontiguousarray(X_train[idx], dtype=np.float32)
ytr = y_train[idx].astype(int)
del X_train, y_train; gc.collect()
print("train:", Xtr.shape, np.bincount(ytr))

dtrain = lgb.Dataset(Xtr, label=ytr, free_raw_data=True)
del Xtr; gc.collect()

params = {"objective": "binary", "num_leaves": 256, "learning_rate": 0.05,
          "num_threads": os.cpu_count(), "verbose": -1, "seed": 42}
model = lgb.train(params, dtrain, num_boost_round=300)
del dtrain; gc.collect()
print(f"[{(time.time()-t0)/60:.1f} min] entraînement terminé")

# --- Évaluation
X_test, y_test = thrember.read_vectorized_features(DATA, subset="test")
it = np.where((y_test == 0) | (y_test == 1))[0]
Xte = np.ascontiguousarray(X_test[it], dtype=np.float32); yte = y_test[it].astype(int)
del X_test, y_test; gc.collect()

p = model.predict(Xte)
del Xte; gc.collect()

fpr, tpr, thr = roc_curve(yte, p)
seuil = float(thr[np.where(fpr <= 0.02)[0][-1]])
auc = float(roc_auc_score(yte, p))

print("\n--- Seuil 0.5 par défaut ---")
print(classification_report(yte, (p >= 0.5).astype(int),
                            target_names=['benin','malveillant'], digits=4))
print("ROC-AUC:", auc)

cm = confusion_matrix(yte, (p >= seuil).astype(int))
fp_rate, det = cm[0,1]/cm[0].sum(), cm[1,1]/cm[1].sum()
print(f"\n--- Seuil optimisé {seuil:.4f} ---\n{cm}")
print(f"Faux positifs : {fp_rate*100:.2f}%  |  Détection : {det*100:.2f}%")

# --- Challenge set
X_chal, y_chal = thrember.read_vectorized_features(DATA, subset="challenge")
p_chal = model.predict(np.ascontiguousarray(X_chal, dtype=np.float32))
det_chal = float((p_chal >= seuil).mean())
print(f"Détection challenge set (malwares évasifs) : {det_chal*100:.2f}%")

# --- Sauvegarde
model.save_model(f"{OUT}/lightgbm_ember2024_v1.txt")
joblib.dump(model, f"{OUT}/lightgbm_ember2024_v1.joblib")
json.dump({"seuil": seuil, "roc_auc": auc, "faux_positifs": float(fp_rate),
           "detection": float(det), "detection_challenge": det_chal,
           "n_features": 2568, "n_train": 200000},
          open(f"{OUT}/metrics_ember2024_v1.json","w"), indent=2)
!ls -lh /content/drive/MyDrive/GuardianAI/models/
print(f"\nTOTAL : {(time.time()-t0)/60:.1f} min")
```

```python
import os, gc, time, json, joblib
import numpy as np, lightgbm as lgb
from sklearn.metrics import roc_curve, roc_auc_score, confusion_matrix

DATA, OUT, DIM = "/content/ember2024", "/content/drive/MyDrive/GuardianAI/models", 2568
assert os.path.exists(f"{DATA}/X_train.dat"), "Donnees absentes : la VM a ete recyclee"
t0 = time.time()

def load(name):
    X = np.memmap(f"{DATA}/X_{name}.dat", dtype=np.float32, mode="r").reshape(-1, DIM)
    y = np.memmap(f"{DATA}/y_{name}.dat", dtype=np.float32, mode="r")
    return X, y

# --- Entrainement
X_train, y_train = load("train")
print("fichier train :", X_train.shape)
yt = np.asarray(y_train)
idx = np.sort(np.random.default_rng(42).choice(np.where((yt == 0) | (yt == 1))[0],
                                               150000, replace=False))
Xtr = np.empty((len(idx), DIM), dtype=np.float32)
for i in range(0, len(idx), 10000):
    Xtr[i:i+10000] = X_train[idx[i:i+10000]]
ytr = yt[idx].astype(np.int8)
del X_train, y_train, yt; gc.collect()
print("train:", Xtr.shape, np.bincount(ytr))

dtrain = lgb.Dataset(Xtr, label=ytr, free_raw_data=True); del Xtr; gc.collect()
params = {"objective":"binary","num_leaves":256,"learning_rate":0.05,
          "num_threads":os.cpu_count(),"verbose":-1,"seed":42}
model = lgb.train(params, dtrain, num_boost_round=400); del dtrain; gc.collect()

joblib.dump(model, f"{OUT}/lightgbm_ember2024_v2.joblib")
print(f"[{(time.time()-t0)/60:.1f} min] modele entraine ET sauvegarde")

# --- Evaluation par lots
X_test, y_test = load("test")
yv = np.asarray(y_test)
it = np.where((yv == 0) | (yv == 1))[0]
yte = yv[it].astype(np.int8)
p = np.empty(len(it))
for i in range(0, len(it), 20000):
    p[i:i+20000] = model.predict(np.asarray(X_test[it[i:i+20000]]))
del X_test, y_test, yv; gc.collect()

fpr, tpr, thr = roc_curve(yte, p)
seuil = float(thr[np.where(fpr <= 0.02)[0][-1]]); auc = float(roc_auc_score(yte, p))
cm = confusion_matrix(yte, (p >= seuil).astype(int))
fp_rate, det = cm[0,1]/cm[0].sum(), cm[1,1]/cm[1].sum()
print(f"\nROC-AUC {auc:.4f} | seuil {seuil:.4f}\n{cm}")
print(f"Faux positifs {fp_rate*100:.2f}% | Detection {det*100:.2f}%")

X_chal, _ = load("challenge")
pc = np.empty(X_chal.shape[0])
for i in range(0, X_chal.shape[0], 20000):
    pc[i:i+20000] = model.predict(np.asarray(X_chal[i:i+20000]))
det_chal = float((pc >= seuil).mean())
print(f"Detection evasifs {det_chal*100:.2f}%")

json.dump({"seuil":seuil,"roc_auc":auc,"faux_positifs":float(fp_rate),
           "detection":float(det),"detection_challenge":det_chal,
           "dataset":"EMBER2024 (.NET + Win64)","n_train":150000},
          open(f"{OUT}/metrics_ember2024_v2.json","w"), indent=2)
!ls -lh /content/drive/MyDrive/GuardianAI/models/
print(f"TOTAL {(time.time()-t0)/60:.1f} min")
```
